```python
"""
Phase 3 -  FEATURE ENGINEERING
— Nassau Candy Factory Reallocation & Shipping Optimization Recommendation System.
Project requirement: predicting shipping Lead Time using Product, Origin Factory, Destination Region, and Ship Mode.Run this first. It reads the raw supplied dataset and writes a cleaned,
feature-engineered version that every later phase reads from.
 
Usage:
    STEP 3 — FEATURE ENGINEERING
"""
```

```python
# Step 3.1 — Load the EDA-ready dataset

import pandas as pd
import numpy as np
from pathlib import Path

# Load the cleaned dataset created in Step 1
file_path = Path(
    "nassau-candy-factory-optimization/data/Nassau_Candy_Cleaned_Step.csv"
)

df = pd.read_csv(file_path)

print("Step 3: Feature Engineering")
print("-" * 50)
print("Dataset shape:", df.shape)
print("\nColumns available:")
print(df.columns.tolist())

display(df.head())
```

```python
# Step 3.2 — Check the columns

required_columns = [
    "Order Date",
    "Ship Date",
    "Ship Mode",
    "Region",
    "Division",
    "Product ID",
    "Product Name",
    "Sales",
    "Units",
    "Gross Profit",
    "Cost",
    "Shipping Lead Time Days"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("All required columns are available.")
```

```python
# Step 3.3 Create date-based features

# Convert date columns to datetime
# Use dayfirst=True only if the original dates are DD-MM-YYYY.
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce",
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce",
    dayfirst=True
)

# Create order-date features
df["Order Year"] = df["Order Date"].dt.year

df["Order Month"] = df["Order Date"].dt.month

df["Order Quarter"] = (
    df["Order Date"].dt.quarter
)

df["Order Day of Week"] = (
    df["Order Date"].dt.dayofweek
)

df["Order Day of Month"] = (
    df["Order Date"].dt.day
)

# Display the newly created features
date_features = [
    "Order Date",
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order Day of Week",
    "Order Day of Month"
]

display(df[date_features].head(10))

print("Date features created successfully.")

# Interpretation: These features allow the model to investigate whether shipping lead time varies by month, quarter, or day of the week.
```

```python
# Cell 4 — Check categorical features



categorical_columns = [
    "Ship Mode",
    "Region",
    "Division",
    "Product ID",
    "Product Name"
]

for col in categorical_columns:
    print(f"\n{col}")
    print("Unique categories:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(10))
```

```python
# Cell 5 — Select features for predicting shipping lead time

#Important decisions
#Exclude Ship Date from the predictors. It is used to calculate the target and is not known when an order is placed.
#Exclude Shipping Lead Time Days from the predictors. It is the value we want to predict.
#Exclude Order ID and Row ID. They are identifiers rather than meaningful operational features.
#Exclude Gross Profit, Cost, and Sales from the initial model. These values may not all be known at the intended prediction time, and using them without validating their availability could introduce leakage.
#Exclude Customer ID initially. It can encourage memorization of individual customers rather than generalizable shipping patterns.

#For now, we will retain Product Name in the cleaned dataset but use Product ID as the initial product feature.
```

```python
# Cell 6 — Create the initial modeling dataset

# Define the target
target_column = "Shipping Lead Time Days"

# Define initial predictor columns
feature_columns = [
    "Region",
    "Ship Mode",
    "Division",
    "Product ID",
    "Units",
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order Day of Week",
    "Order Day of Month"
]

# Check that all features and the target exist
required_model_columns = feature_columns + [target_column]

missing_model_columns = [
    col for col in required_model_columns
    if col not in df.columns
]

if missing_model_columns:
    raise ValueError(
        f"Missing modeling columns: {missing_model_columns}"
    )

# Create a separate modeling dataset
model_df = df[required_model_columns].copy()

print("Initial modeling dataset created.")
print("Shape:", model_df.shape)

display(model_df.head())
```

```python
# Cell 7 — Check missing values and invalid targets


print("Missing values by column:")
display(
    model_df.isnull().sum().to_frame(
        name="Missing Values"
    )
)

print("\nTarget variable summary:")
display(
    model_df[target_column].describe()
)

print(
    "\nNegative shipping lead times:",
    (model_df[target_column] < 0).sum()
)

print(
    "Zero-day shipping lead times:",
    (model_df[target_column] == 0).sum()
)
```

```python
# Cell 8 — Prepare the data for modeling

# Copy the initial modeling dataset
model_df = model_df.copy()

# Remove rows without a target
model_df = model_df.dropna(
    subset=[target_column]
)

# Separate the predictors and target
X = model_df[feature_columns].copy()
y = model_df[target_column].copy()

print("Predictor matrix shape:", X.shape)
print("Target vector shape:", y.shape)

print("\nPredictor data types:")
display(X.dtypes.to_frame(name="Data Type"))

print("\nTarget summary:")
display(y.describe())
```

```python
# Cell 9 — Save the feature-engineered dataset

# Save the full feature-engineered dataset
output_file = Path(
    "nassau-candy-factory-optimization/data/Nassau_Candy_Feature_Engineered_Step3.csv"
)

output_file.parent.mkdir(
    parents=True,
    exist_ok=True
)

df.to_csv(
    output_file,
    index=False
)

print("Feature-engineered dataset saved successfully.")
print("File location:", output_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Cell 10 — Verify the saved dataset

df_step3 = pd.read_csv(output_file)

print("Saved dataset loaded successfully.")
print("Shape:", df_step3.shape)

print("\nNew features:")
new_features = [
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order Day of Week",
    "Order Day of Month"
]

display(df_step3[new_features].head())

assert "Shipping Lead Time Days" in df_step3.columns
assert all(col in df_step3.columns for col in new_features)

print("Verification passed.")
```
